Empezaremos importando los datos ya limpios y listos para entrenamiento.

```python
import pandas as pd
from pathlib import Path

path = Path(
  r"C:\Users\antim\OneDrive\Escritorio\poryectos\proyecto_videojuegos\data\data_set_ml_01.parquet"
)

data = pd.read_parquet(path)
```

```python
data.head()
```

## 1. Split de datos.
Luego de tener nuestros datos cargados podemos dividirlos en la matriz independiente `X` y la variable depnediente `y` .

```python
# Definiomos a X como todas las columnas menos la que sera y
X = data.drop(columns='global_sales')
# Asignamos a y nuestra variable dpendeiente
y = data['global_sales']

print(f"Shape de 'X': {X.shape}\nShape de 'y': {y.shape}")
```

Ya teniendo separados nuestros datos podemos hacer el train/test split. Para esto importaremos el modulo `train_test_split`. 

Uitilizaremos un 20% del dataset en el testset

```python
# importamos dependencias
from sklearn.model_selection import train_test_split

X_trian, X_test, y_train, y_test = train_test_split(
  X, 
  y, 
  test_size= 0.20,
  random_state=42
)

print(
  f"Size del test/train de 'X': {X_test.shape}/{X_trian.shape}\nSize del test/train de 'y': {y_test.shape}/{y_train.shape}"
)
```

## 2. Instanciamos el RandomForest

Esta vez utilizarmos el modelo `RandomForestRegressor`, en este caso nos es muy util porque el RandomForest nos permite hacer regresiones no lineales.

Es importante seleccionar los hiperparametros `n_jobs=-1`, ya que esto nos permite utilizar todos los nucleos de la cpu.

```python
# importamos el modulo
from sklearn.ensemble import RandomForestRegressor

# n_estimators :cant de arboles, n_jobs: utiliza todos los nucleos del procesador
modelo_random_forest_01 = RandomForestRegressor(n_estimators=100, n_jobs=-1, random_state=42)
```

## 3. Entrenamiento del modelo

Entrenamos el modelo con el metodo `.fit(X_train, y_train)`

```python
# Midamos el tiempo de ejecucion con la sentencia %time
# Utilizamos el metodo .fit
%time modelo_random_forest_01.fit(X_trian, y_train)
```

## 4. Hagamos predicciones

Utilizaremos el metodo `predict(X_test)`, para que el modelo genere predicciones que luego compararemos con los valores reales.

```python
# Le asiganmos a las variabe y_preds las predicciones hechas por el modelo
y_pred = modelo_random_forest_01.predict(X_test)
```

## 5. Calculamos las metricas
Una vez que realizamos las predicciones, podemos medir que tan distantes estan los valores dados por el modelo de los valores reales, esto lo haremos utilizando dos metricas como el `mae` y el `r2`.

```python
# importamos las dependencias 
from sklearn.metrics import mean_absolute_error, r2_score

mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"\n--- REPORTE DE DESEMPEÑO ---")
print(f"Error Absoluto Medio (MAE): {mae:.4f}")
print(f"Coeficiente de determinación (^): {r2:.4f}")
```
