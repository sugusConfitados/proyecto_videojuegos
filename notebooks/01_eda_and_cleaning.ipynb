# Limpieza de datos

Ahora limpiaremos los datos y luego realizaremos el EDA Estrategico.

Limpieza : 

0. Columnas a lowerCase:
    Antes de comenzar con la limpieza de datos en si, primero limpiaremos las columnas haceindo que estas sean minuscula en vez de mayuscula.

1. Gestion de datos nulos en la columna `'year'`
    Lo que haremos sera eliminar los datos nulos de la columna 'year', ya que la cantidad de datos nulos solo reepresenta el 2%, y en estos casos es mejor que los datos tengan menos ruido y no cambiarlos por valores como la media o mediana. Para esto utilizaremos el metodo `.dropna()`.\

2. Intercambio de datos nulo en la columna `'publisher'`
    Cambiaremos los datos nulos de la coumna 'publisher' con el string 'Unknown', para no perder las filas que no tienen publisher. Esto lo haremos con el metodo `.fillna()`

3. Duplicados
    Eliminaremos aquellos registros duplicados donde el name y la plataform sea la misma. Para esto utilizaremos el metodo `.drop_duplicates(subset=[])`

4. Casting de tipos de dato
    Una vez eliminemos lo registros con year nulo, podemos hacer un casteo de la columna year podemos pasar la columna de tipo float a int.

```python
# cambiamos el atributo publico columns a lower(), !!! importante utilizar el .str para wue s3ea un string
data.columns = data.columns.str.lower()

data.columns
```

ya tenemos las columnas en minuscula sigamos con la getsion de nulos con la columna 'year'.

Utilizaremos el metodo `dropna()` --> recordar que este metodo es un metodo inmutable por lo que no modifica el dataframe original.

```python
# primero veamos la cantidad de valores nulos que tenemos en la columna year
print(data['year'].isnull().sum())

# eliminamos los valores nulos de la columna
data = data.dropna(subset="year")

print(data['year'].isnull().sum())
```

Vemos como pasamos de 271 valores nulos a 0. 

Ahora lo que haremos sera rempplzar los datos nulos de la columna 'publisher' por el str 'Unknown'

```python
# primero veamos los valores nulos de la columna 
print(data['publisher'].isnull().sum())

# cambiamos los valores nulos por 'Unknown'
data['publisher'] = data['publisher'].fillna('Unknown') # es importaante seleccionar que columna queremos remplzar

# veamos que valores nulos quedaron
print(data['publisher'].isnull().sum())
```

Ahora vemos como no tenemos mas valores nulos dentro de la columna 'publisher', sino que ahora aparecera el dato 'Unkonwn'

Ahora eliminaremos los registros dupliocados , estos los identiicaremos si su columna 'name' y 'plataaform' son iguales, nos quedaremos con el primer registro encontrado, eliminando todos los demas.

```python
# primero contemos cuantos registros tienen al menos una concidencia
print(f"Cantidad de duplicados: {data[['name', 'platform']].duplicated().sum()}")
# contemos cuantos duplicados tienen estos registros
print(f"\nJuegos y plataformas duplicados:\n{data[['name', 'platform']].value_counts()[data[['name', 'platform']].value_counts() > 1]}\n")

# eliminamos los duplicados utlizando el drop_duplicated -> es un metodo inmutable, por lo que hay que hacer una asiganacion
data = data.drop_duplicates(subset=['name', 'platform'], keep='first') # utilizamos el keep para definir con que registro nos queremos quedar

# imprimammos cuantos quedaron
print(f"Cantidad de duplicados: {data[['name', 'platform']].duplicated().sum()}")
```

Ahora cambiaremos el tipo de dato de las columnas utilizando el metodo `astype`.

```python
data['year'] = data['year'].astype(int)

data['year'].dtype
```

Por timo veamos si tenemos nulos silenciosos en alguno de nuestros registros en la columna 'name'

```python
# Esto busca filas donde el nombre, tras quitar espacios, quede vacío
nombres_vacios = data[data['name'].str.strip() == ""]
print(f"Nombres que son solo espacios: {len(nombres_vacios)}")
```

Podemos ver que no tenemos valores de name que sean nulos silenciosos.
